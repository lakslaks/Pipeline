```python
steps = [
    {"name": "always", "action": "builtins#print", "args": ["Always runs"]},
    {"name": "only_if_true", "action": "builtins#print", "args": ["x > 10 triggered!"], "when": "${x} > 10"}
]

executor = PipelineExecutor(name="conditional", steps=steps)
executor.context["x"] = 15  # Try changing to 5 to skip
asyncio.run(executor.execute())
```

```python
def before(name, step, ctx):
    print(f"==> Starting step: {name}")

def after(name, step, ctx, result):
    print(f"<== Finished step: {name} with result: {result}")

steps = [
    {"name": "hooked", "action": "builtins#print", "args": ["Inside hooked step."]},
    {"name": "always", "action": "builtins#print", "args": ["Always runs"]},
]

executor = PipelineExecutor(name="hooks", steps=steps, execution_controllers={
    "before": before,
    "after": after
})
asyncio.run(executor.execute())
```
